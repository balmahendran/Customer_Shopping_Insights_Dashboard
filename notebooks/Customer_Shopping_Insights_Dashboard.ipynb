### Import Libraries

```python
import pandas as pd
import numpy as np    
```

### Data Loading

```python
DF = pd.read_csv('../Raw Data/customer_shopping_behavior.csv')
```

### Data Analysis

```python
DF.info()
DF.describe(include='all')
```

### Column name Change

```python
DF.columns = DF.columns.str.replace(' ', '_').str.lower()
DF.columns
DF.columns = DF.columns.str.replace('purchase_amount_(usd)', 'purchase_amount').str.lower()
DF.columns
```

### Null Value Handle

```python
DF.isnull().sum()
DF['review_rating'] = DF.groupby('category')['review_rating'].transform(lambda x: x.fillna(x.median()))
DF.isnull().sum()
```

### Create Age Group Column

```python
DF['age'].head()
bins = [0,19, 30, 60,100]  
labels = ['Teenager', 'Adult','Middle-Aged', 'Senior']
DF['age_group'] = pd.cut(DF['age'], bins=bins, labels=labels)
DF.groupby('age_group')['age'].unique()
```

### Convert Frequency Purchase to Days in No

```python
DF['frequency_of_purchases'].unique()
freq_days_map = {
    'Weekly': 7,
    'Fortnightly': 14,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Every 3 Months': 90,
    'Annually': 365
}
DF['purchase_cycle_days'] = DF['frequency_of_purchases'].map(freq_days_map)
DF.groupby('frequency_of_purchases')['purchase_cycle_days'].unique()
```
